```python
import nbformat
# Load your notebook

fix_list_name = ["Llama2_LoRA_Project", "NER_with_BERT_Project", "Q&A_with_LM", "Sentence_Embedding_Project", "Text_Cleaning_Project", "Text_Summarization_Project", "Word2Vec_Project"]

for file_name in fix_list_name:

    nb = nbformat.read(f"{file_name}.ipynb", as_version=nbformat.NO_CONVERT)

    # Add the missing state key if widgets exist
    if "widgets" in nb.metadata and "state" not in nb.metadata.widgets:
        nb.metadata["widgets"] = {"state": {}}

    # Save the fixed version
    nbformat.write(nb, f"{file_name}.ipynb")
```
