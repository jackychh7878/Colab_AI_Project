```python
# Strip ipywidget outputs so GitHub can render notebooks (keeps text/plain results).
# Same logic as fix_notebooks_for_github.py in this folder — run that script from the repo root for all .ipynb files.

import nbformat

WIDGET_VIEW = "application/vnd.jupyter.widget-view+json"

fix_list_name = [
    "Llama2_LoRA_Project", "NER_with_BERT_Project", "Q&A_with_LM",
    "Sentence_Embedding_Project", "Text_Cleaning_Project",
    "Text_Summarization_Project", "Word2Vec_Project",
]

for file_name in fix_list_name:
    path = f"{file_name}.ipynb"
    nb = nbformat.read(path, as_version=nbformat.NO_CONVERT)
    for cell in nb.cells:
        if cell.get("cell_type") != "code":
            continue
        outputs = cell.get("outputs") or []
        new_outputs = []
        for out in outputs:
            if out.get("output_type") in ("display_data", "execute_result"):
                data = out.get("data")
                if isinstance(data, dict) and WIDGET_VIEW in data:
                    del data[WIDGET_VIEW]
                if isinstance(data, dict) and not data:
                    continue
            new_outputs.append(out)
        cell["outputs"] = new_outputs
    if "widgets" in nb.metadata:
        del nb.metadata["widgets"]
    nbformat.write(nb, path)
    print("fixed:", path)
```
